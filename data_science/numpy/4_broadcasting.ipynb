```python
import numpy as np

A = np.arange(9.).reshape(3,3)   # 2d array : (3,3)
x = np.array([1.,0,0])   # 1d array : (3,)
y = x.reshape(1,3)       # 2d array : (1,3)
z = x.reshape(3,1)       # 2d array : (3,1)

print(A)
print(x)
print(y)
print(z)
```

```python
print(A + 1)
print(A + x)
print(A + y)
print(A + z)
```
