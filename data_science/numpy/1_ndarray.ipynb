```python
# ndarray
# https://wikidocs.net/14606
import numpy as np

x = np.array((0.1,0.2,0.3))    # np.array([0.1,0.2,0.3])도 가능

print(x)
print(x.shape)
print(x.dtype)

y = np.array(((1,2,3),(4,5,6)))    # [(1,2,3),(4,5,6)],[[1,2,3],[4,5,6]] 등도 가능

print(y)
print(y.dtype)
print(y.shape)

z = np.array([1,2,3],dtype='float64')

print(z)
print(z.dtype)
print(z.shape)
```

```python
X = np.zeros((3,3))
print(X)

Y = np.ones((3,3),dtype='int32')
print(Y)

Z = np.empty((3,3))
print(Z)
```

```python
import numpy as np

A = np.array([])
for i in range(3):
    A = np.append(A,[1,2,3])

print(A)
```

```python
print(np.arange(1,2,0.1))
print(np.arange(10))   # start, step 생략가능. 정수로 생성)
print(np.arange(10.))  # start, step 생략가능. 실수로 생성)
print(np.linspace(0.,20.,11))
print(np.eye(3))
```

```python
X = np.arange(0,9,1.)
print(X)

Y = np.reshape(X,(3,3))
Z = X.reshape((3,3))
print(Y)
print(Z)

X.shape = (3,3)
print(X)
```

```python
a = np.arange(3);
a = a.astype(int)
a = a.astype('int32')
print(a.dtype)
a = a.astype('int64')
print(a.dtype)

a = a.astype(float)    # a.astype('float64')
a = a.astype('float32')
print(a.dtype)
a = a.astype('float64')
print(a.dtype)
```
