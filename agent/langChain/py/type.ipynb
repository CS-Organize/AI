# typing

## Annotated

`Annotated[T, x, ...]`

- `T` 타입에 대한 메타데이터 `x`를 추가
- 라이브러리 또는 도구가 metadata를 정적 분석 도구 또는 런타임에 사용할 수 있음


- https://docs.python.org/3/library/typing.html#typing.Annotated
- https://stackoverflow.com/questions/71898644/how-to-use-python-typing-annotated

```python
from dataclasses import dataclass
from typing import Annotated, get_origin

@dataclass
class ValueRange:
    min: int
    max: int

def metafunc():
    return "This is metadata"

a = Annotated[int, "This is an integer", ValueRange(0, 10), metafunc]
print(a)
print(a.__origin__)
print(a.__metadata__)
print(get_origin(a))
```

## TypedDict

타입스크립트의 `interface`와 유사한 기능을 제공

- https://docs.python.org/3/library/typing.html#typing.TypedDict

```python
from typing import TypedDict

class Point2D(TypedDict):
    x: int
    y: int
    label: str

a: Point2D = {'x': 1, 'y': 2, 'label': 'good'} # ok
print(a)
b: Point2D = {'z': 3, 'label': 'bad'} # fails type check
print(b)

# https://wikidocs.net/21043
try:
    assert Point2D(x=1, y=2, label='good') == {'x': 1, 'y': 2, 'label': 'good'}
    assert Point2D(x=1, y=2, label='good') == {'x': 1, 'y': 2, 'label': 'good', 'z': 3}
    # a = 1 / 0
except (AssertionError, ZeroDivisionError) as e:
    print(f"error1: {e if isinstance(e, ZeroDivisionError) else 'assertion failed'}")
finally:
    print("done")

class InheritPoint3D(Point2D):
    z: int

class Point3D(TypedDict):
    x: int
    y: int
    z: int
    label: str

try:
    assert InheritPoint3D(x=1, y=2, z=3, label='good') == {'x': 1, 'y': 2, 'z': 3, 'label': 'good'}
    assert Point3D(x=1, y=2, z=3, label='good') == {'x': 1, 'y': 2, 'z': 3, 'label': 'good'}
except AssertionError as e:
    print(f"error2: {e}")

# total=False -> 모든 필드를 강제하지 않음
class TotalFalse(TypedDict, total=False):
    x: int
    y: int

class TotalTrue(TypedDict, total=True):
    x: int
    y: int

c = TotalFalse(x=1) # ok
print(c)
d = TotalTrue(x=1) # fails type check
print(d)
```
