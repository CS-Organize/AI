```python
from langchain_ollama import ChatOllama

reasoning_llm = ChatOllama(model="jinbora/deepseek-r1-Bllossom:8b", stop=["</think>"])
answer_llm = ChatOllama(model="exaone3.5", temperature=0)
```

```python
from typing_extensions import Annotated, List, TypedDict, Literal

from langchain_core.documents import Document
from langgraph.graph.message import add_messages


class RAGState(TypedDict):
    query: str
    thinking: str
    documents: List[Document]
    answer: str
    messages: Annotated[List, add_messages]
    mode: Literal["retrieve", "generate"]
```

```python
from langchain_docling.loader import DoclingLoader, ExportType

FILE_PATH = "https://arxiv.org/pdf/2408.09869"

loader = DoclingLoader(file_path=FILE_PATH, export_type=ExportType.MARKDOWN)
docs = loader.load()
```

```python
from IPython.display import display, Markdown

display(Markdown(docs[0].page_content))
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "Header_1"),
        ("##", "Header_2"),
        ("###", "Header_3"),
    ]
)

splits = [split for doc in docs for split in splitter.split_text(doc.page_content)]

for d in splits[:3]:
    print(f"- {d.page_content=}")
print("...")
```

```python
from IPython.display import display, Markdown

display(Markdown(splits[12].page_content))
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(
    model="bge-m3:latest",
)
```

```python
from langchain_qdrant import QdrantVectorStore, RetrievalMode

vector_store = QdrantVectorStore.from_documents(
    documents=splits,
    embedding=embeddings,
    location=":memory:",
    collection_name="rag_collection_0315",
    retrieval_mode=RetrievalMode.DENSE,
)

retriever = vector_store.as_retriever(search_kwargs = {"k":10})
```

```python
from langchain.retrievers import ContextualCompressionRetriever
from langchain.retrievers.document_compressors.cross_encoder_rerank import CrossEncoderReranker
from langchain_community.cross_encoders import HuggingFaceCrossEncoder

model = HuggingFaceCrossEncoder(model_name="BAAI/bge-reranker-base")

compressor = CrossEncoderReranker(model=model, top_n = 5)
compressor_retriever = ContextualCompressionRetriever(
    base_compressor=compressor,
    base_retriever=retriever,
)
```

```python
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

from langgraph.graph import START, StateGraph, END

def classify_node(state: RAGState):
    query = state["query"]

    if "Docling" in query:
        print("검색 시작")
        return {"mode": "retrieve"}
    else:
        print("생성 시작")
        return {"mode": "generate"}

def route_by_mode(state: RAGState) -> Literal["retrieve", "generate"]:
    return state["mode"]

def retrieve(state: RAGState):
    query = state["query"]

    print("검색 중")
    documents = compressor_retriever.invoke(query)
    for doc in documents:
        print(f"- {doc.page_content=}")
    print("검색 완료")

    return {
        "documents": documents,
    }


def reasoning(state: RAGState):
    query = state["query"]
    documents = state["documents"]
    context = "\n\n".join([doc.page_content for doc in documents])
    reasoning_prompt = ChatPromptTemplate.from_template(
        """주어진 문서를 활용하여 사용자의 질문에 가장 적절한 답변을 작성해주세요.
질문: {query}

문서 내용:
{context}

상세 추론:""",)

    reasoning_chain = reasoning_prompt | reasoning_llm | StrOutputParser()
    print("추론 중")
    thinking = reasoning_chain.invoke({"query": query, "context": context})

    return {
        "thinking": thinking,
    }

def generate(state: RAGState):
    query = state["query"]
    thinking = state["thinking"]
    documents = state["documents"]

    context = "\n\n".join([doc.page_content for doc in documents])

    answer_prompt = ChatPromptTemplate.from_template(
        """사용자 질문에 한글로 답변하세요. 제공된 문서와 추론 과정이 있다면, 최대한 활용하세요.

질문: {query}

추론과정:
{thinking}

문서 내용:
{context}

답변:""",)

    print("답변 중")
    answer_chain = answer_prompt | answer_llm | StrOutputParser()
    answer = answer_chain.invoke({"query": query, "thinking": thinking, "context": context})
    print("답변 완료")
    return {
        "answer": answer,
        "messages": [HumanMessage(content=answer)],
    }
```

```python
from langgraph.checkpoint.memory import MemorySaver

builder = StateGraph(RAGState)

builder.add_node("classify", classify_node)
builder.add_node("reasoning", reasoning)
builder.add_node("retrieve", retrieve)
builder.add_node("generate", generate)

builder.add_edge(START, "classify")
builder.add_conditional_edges(
    "classify",
    route_by_mode,
    {
        "retrieve": "retrieve",
        "generate": "generate",
    }
)
builder.add_edge("retrieve", "reasoning")
builder.add_edge("reasoning", "generate")
builder.add_edge("generate", END)

memory = MemorySaver()
graph = builder.compile(checkpointer=memory)
```

```python
from IPython.display import Image, display

display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
input = {"query": "Docling이 무엇인가요?"}
config = {"configurable": {"thread_id": 0}}

output = graph.invoke(input, config=config)
print(output)
```
